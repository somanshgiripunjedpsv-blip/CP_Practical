```python
# If else statement for marks of the students
# 95 and above - grade A, 85-94 - grade C, 75-84 - grade D, 65-74 - grade B, below 65 - grade F
marks = int(input("Enter your marks: "))
if marks >= 95:
    print("You got grade A")
elif marks >= 85:
    print("You got grade C")
elif marks >= 75:
    print("You got grade D")
elif marks >= 65:
    print("You got grade B")
else:
    print("You got grade F")
```
