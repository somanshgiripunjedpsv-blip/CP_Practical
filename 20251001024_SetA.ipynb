```python

try:
    # (a) Read equilibrium constant
    raw_input_value = input("Enter the value of Kc: ")

    print("Data type of raw input:", type(raw_input_value))

    # Convert input into float
    Kc = float(raw_input_value)

    print("Data type after conversion:", type(Kc))

    # Check whether Kc is positive
    if Kc <= 0:
        print("Error: Kc must be greater than zero.")

    else:
        # (b) Calculate quadratic coefficients
        a = Kc
        b = -(2 * Kc + 1)
        c = Kc

        print("\nQuadratic equation:")
        print(f"{a}x^2 + ({b})x + {c} = 0")

        # Calculate discriminant
        D = b**2 - 4*a*c

        print("\nCoefficient a =", a)
        print("Coefficient b =", b)
        print("Coefficient c =", c)
        print("Discriminant =", D)

        # Calculate both roots
        if D >= 0:
            root1 = (-b + D**0.5) / (2*a)
            root2 = (-b - D**0.5) / (2*a)

            print("Root 1 =", root1)
            print("Root 2 =", root2)

            # (c) Select physically meaningful root
            if 0 < root1 < 1:
                x = root1
            elif 0 < root2 < 1:
                x = root2
            else:
                x = None

            if x is None:
                print("No physically meaningful root found.")

            else:
                # (d) Calculate equilibrium concentrations
                A = 1 - x
                B = 1 - x
                C = x

                print("\nEquilibrium concentrations:")
                print("[A] =", A, "mol/L")
                print("[B] =", B, "mol/L")
                print("[C] =", C, "mol/L")

                # Conversion of limiting reactant
                conversion = (x / 1) * 100

                print("Conversion of limiting reactant =",
                      conversion, "%")

                # (e) Classify conversion
                print("\nConversion classification:")

                if conversion >= 70:
                    print("High conversion")
                elif conversion >= 40:
                    print("Moderate conversion")
                else:
                    print("Low conversion")

        else:
            print("Error: Discriminant is negative.")

except ValueError:
    print("Error: Please enter a valid numerical value for Kc.")


```

```python

import math

# Given data
C0 = 2.0
k = 0.08
dt = 1.0

# Part (i): Euler method using for loop
C = C0

print("Part (i): Euler Method")
print("---------------------------------------------")
print("Time(min)   Concentration(mol/L)   Conversion(%)")

for t in range(11):
    conversion = ((C0 - C) / C0) * 100
    print(f"{t:<11} {C:<23.4f} {conversion:.2f}")

    C = C - k * C * dt

# Part (ii): Exact solution and absolute error
C_euler = C
C_exact = C0 * math.exp(-k * 10)

error = abs(C_exact - C_euler)

print("\nPart (ii): Exact Solution")
print("Euler concentration =", round(C_euler, 4), "mol/L")
print("Exact concentration =", round(C_exact, 4), "mol/L")
print("Absolute error =", round(error, 4), "mol/L")

# Part (iii): While loop until conversion reaches 60%
C = C0
time = 0

while ((C0 - C) / C0) < 0.60 and time < 100:
    C = C - k * C * dt
    time += dt

print("\nPart (iii): While Loop")
print("Time needed =", time, "min")
print("Final concentration =", round(C, 4), "mol/L")
print("Final conversion =", round((C0-C)/C0*100, 2), "%")
```

```python
def factorial(n):
    if n < 0:
        return None
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0 or r > n:
        return "Invalid input"

    return factorial(n) // (factorial(r) * factorial(n-r))


# Test cases
print("Combination (6,3):", combination(6, 3))
print("Combination (5,7):", combination(5, 7))
```

```python

def fib_iter(n):
    fib = []
    a, b = 0, 1

    for i in range(n):
        fib.append(a)
        a, b = b, a + b

    return fib


def fib_rec(k):
    if k == 0:
        return 0
    elif k == 1:
        return 1
    else:
        return fib_rec(k - 1) + fib_rec(k - 2)


# Number of samples
n = 10

# Generate Fibonacci numbers iteratively
fib_list = fib_iter(n)

# Check both functions agree
for i in range(n):
    assert fib_list[i] == fib_rec(i)

print("Both functions agree!")

# Print sampling table
print("\nSample No. | Fibonacci tag | Sampling time (min)")
print("-" * 52)

for i in range(n):
    sampling_time = fib_list[i] * 2
    print(f"{i+1:^10} | {fib_list[i]:^14} | {sampling_time:^20}")
```
